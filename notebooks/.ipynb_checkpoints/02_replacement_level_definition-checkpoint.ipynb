```python
import os
import pandas as pd

PROC_DIR = "../data/processed"
IN_PATH = os.path.join(PROC_DIR, "darko_with_minutes_2025_26.csv")

df = pd.read_csv(IN_PATH)
df.shape, df.columns
```

```python
replacement_dpm = df["impact_dpm"].quantile(0.20)
replacement_dpm
```

```python
df["VAR"] = df["impact_dpm"] - replacement_dpm
df[["player_name","team","impact_dpm","VAR","minutes"]].head()
```

```python
df["VAR"].describe()
```

```python
df.sort_values("VAR", ascending=False)[["player_name","team","impact_dpm","VAR","minutes"]].head(15)
```

```python
OUT_PATH = os.path.join(PROC_DIR, "player_var_2025_26.csv")
df.to_csv(OUT_PATH, index=False)
print("Saved:", OUT_PATH)
```
