```python
import os
import pandas as pd

PROC_DIR = "../data/processed"
IN_PATH = os.path.join(PROC_DIR, "player_var_2025_26.csv")

df = pd.read_csv(IN_PATH)
df.shape
```

```python
df["total_value"] = df["VAR"] * df["minutes"]

df[["player_name","team","VAR","minutes","total_value"]].head()
```

```python
df["total_value"].describe()
```

```python
df.sort_values("total_value", ascending=False)[["player_name","team","total_value","VAR","minutes"]].head(15)
```

```python
team_value = (
    df.groupby("team")["total_value"]
      .sum()
      .reset_index()
      .sort_values("total_value", ascending=False)
)

team_value.head(10)
```

```python
df.to_csv(os.path.join(PROC_DIR, "player_total_value_2025_26.csv"), index=False)
team_value.to_csv(os.path.join(PROC_DIR, "team_total_value_2025_26.csv"), index=False)

print("Saved player_total_value_2025_26.csv and team_total_value_2025_26.csv")
```
