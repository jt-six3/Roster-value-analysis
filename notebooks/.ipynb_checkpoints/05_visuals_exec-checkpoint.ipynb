```python
import os
import pandas as pd
import matplotlib.pyplot as plt

PROC_DIR = "../data/processed"
PATH = os.path.join(PROC_DIR, "player_cost_efficiency_2025_26.csv")

df = pd.read_csv(PATH)
df.shape
```

```python
plt.figure(figsize=(10,6))

plt.scatter(df["salary"], df["total_value"], alpha=0.5)

plt.xscale("log")
plt.xlabel("Salary (USD, log scale)")
plt.ylabel("Total Value (VAR × Minutes)")
plt.title("Salary vs Total Value (2025–26 NBA)")

plt.show()
```

```python
team_value = (
    df.groupby("team")["total_value"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10,8))
team_value.plot(kind="barh")

plt.xlabel("Team Total Value (Sum of VAR × Minutes)")
plt.title("Team Total Value (2025–26)")

plt.gca().invert_yaxis()
plt.show()
```

```python
top15 = (
    df.sort_values("value_per_million", ascending=False)
      .head(15)
)

plt.figure(figsize=(10,6))
plt.barh(top15["player_name"], top15["value_per_million"])

plt.xlabel("Value per $1M")
plt.title("Top 15 Value Contracts (≥ 500 minutes)")

plt.gca().invert_yaxis()
plt.show()

top15[["player_name","team","salary","value_per_million"]]
```
